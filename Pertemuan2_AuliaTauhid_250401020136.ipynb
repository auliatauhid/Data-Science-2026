```python
import pandas as pd
from IPython.display import display

# URL alternatif dari Stanford (lebih stabil)
url = "https://web.stanford.edu/class/archive/cs/cs109/cs109.1166/stuff/titanic.csv"
df = pd.read_csv(url)
display(df.head())
```

```python
df.info()
```

```python
print("Shape:", df.shape)
print("Kolom:", df.columns.tolist())
print(df.isnull().sum())
```

```python
wanita_kelas1 = df[(df['Sex'] == 'female') & (df['Pclass'] == 1)]
print("Jumlah yang selamat:", wanita_kelas1['Survived'].sum())
print("Rata-rata usia:", round(wanita_kelas1['Age'].mean(), 2))
```

```python
survival_per_kelas = df.groupby('Pclass')['Survived'].mean()
print(survival_per_kelas.apply(lambda x: f"{x*100:.1f}%"))
print("Kelas tertinggi:", survival_per_kelas.idxmax())
```
